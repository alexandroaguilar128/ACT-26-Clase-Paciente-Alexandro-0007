### Descripcion
🩺 Descripción del Sistema de Gestión de Pacientes
Este código representa una estructura básica y funcional de Programación Orientada a Objetos (POO) en Python, diseñada para ser directa y eficiente. Aquí te explico qué sucede tras bambalinas:

La Clase Paciente 🏗️: Actúa como un "molde" o plantilla maestra. Define que cualquier paciente que registremos debe tener obligatoriamente tres características: nombre, edad y diagnóstico.

Captura de Datos Interactiva 📥: Gracias a la función leerdatos(), el programa se detiene y establece una conversación contigo. Utiliza el teclado para captar la información y guardarla permanentemente en la memoria de cada objeto mediante la función input().

Identidad de Objetos 👤👤: Hemos creado dos "individuos" distintos: paciente1 y paciente2. Lo interesante es que, aunque usan el mismo molde, la información de uno no se mezcla con la del otro; cada uno mantiene su propia identidad digital.

Visualización de Resultados 📋: La función mostrardatos() se encarga de darle un formato limpio y profesional a la salida. Al final del proceso, el código recupera los atributos guardados y los presenta como una ficha médica organizada.

Flujo Secuencial ⚡: Al eliminar bucles (while/for) y decisiones (if), el código sigue una línea recta y predecible. Es una ejecución puramente secuencial: Entrada A ➡️ Entrada B ➡️ Salida A ➡️ Salida B.

![Info](https://raw.githubusercontent.com/alexandroaguilar128/ACT-26-Clase-Paciente-Alexandro-0007/refs/heads/main/Infoclasepaciente.png)

```python
class Paciente:
    def __init__(self, nombre, edad, diagnostico):
        self.nombre = ""
        self.edad = ""
        self.diagnostico = ""

    def leerdatos(self):
        print(f"--- Ingreso de datos ---")
        self.nombre = input("Nombre del paciente: ")
        self.edad = input("Edad del paciente: ")
        self.diagnostico = input("Diagnóstico médico: ")

    def mostrardatos(self):
        print(f"\nFicha del Paciente:")
        print(f"Nombre: {self.nombre}")
        print(f"Edad: {self.edad}")
        print(f"Diagnóstico: {self.diagnostico}")
        print("-" * 20)

# Creación de los 2 objetos
paciente1 = Paciente()
paciente2 = Paciente()

# Ejecución secuencial de funciones
paciente1.leerdatos()
paciente2.leerdatos()

paciente1.mostrardatos()
paciente2.mostrardatos()
print("software realizado por Alexandro Aguilar NL = 1")
```
